```python
import os
import datetime 
```

```python
#FUNCTION DEFINITIONS

#for dealing with the infernal .ipynb_checkpoint files created everywhere    
def listdir_nohidden(path):
    return [f for f in sorted(os.listdir(path)) if not f.startswith('.')]

#MAKE DIRECTORY NAMED WITH A TIMESTAMP AND RETURN IT

#detected_seed_images 
#germination_coordinates
#germination_images
#path_plots - not sure we need this
#final_images_plots
#seed_coordinates
#tip_coordinates
#final_images
#reports
#r_xy_plots
#r_spline_plots
#quantification_summaries- csv with quantification data that can be used for analysis
def prep_dirs(exp_path):
    now = datetime.datetime.now()
    exp_list = listdir_nohidden(exp_path) #listdir_nohidden(exp_path)
    t = min(exp_list) + "_" + max(exp_list) + "_" + str(now.year) + "_" + str(100+now.month)[1:] + "_" + str(100+now.day)[1:] + "_" + str(100+now.hour)[1:] + "_" + str(100+now.minute)[1:] + "_" + str(100+now.second)[1:] 
    dirs = ("detected_seed_images","germination_coordinates","germination_images","path_plots","final_images_plots","seed_coordinates","tip_coordinates","final_images","reports","r_xy_plots","r_spline_plots","quantification_summaries")
    for d in dirs:
        call = "mkdir -p /home/robotmessenger810/file_sorting_quantification/data/quantification_outs/" + t + "/" + d 
        os.system(call)
    return(t)

```
